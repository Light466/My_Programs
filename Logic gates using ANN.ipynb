```python
# ============================================
# Single Layer Perceptron for Logic Gates
# AND, OR, NOT, NAND, NOR
# ============================================

def activation(net):
    """Binary Step Activation Function"""
    if net >= 0:
        return 1
    else:
        return 0


def perceptron(x1, x2, weights, bias):
    net = x1 * weights[0] + x2 * weights[1] + bias
    output = activation(net)
    return net, output


# Logic gate parameters
gates = {
    "AND":  {"weights": [1, 1],  "bias": -1.5},
    "OR":   {"weights": [1, 1],  "bias": -0.5},
    "NAND": {"weights": [-1, -1], "bias": 1.5},
    "NOR":  {"weights": [-1, -1], "bias": 0.5},
}

inputs = [
    (0, 0),
    (0, 1),
    (1, 0),
    (1, 1)
]

print("=" * 72)
print("       SINGLE LAYER PERCEPTRON IMPLEMENTATION OF LOGIC GATES")
print("=" * 72)

for gate, params in gates.items():

    print("\n" + "=" * 72)
    print(f"{gate} GATE")
    print("=" * 72)
    print(f"Weights = {params['weights']}    Bias = {params['bias']}\n")

    print("{:^8} {:^8} {:^15} {:^10}".format(
        "Input1", "Input2", "Weighted Sum", "Output"))
    print("-" * 50)

    for x1, x2 in inputs:
        net, output = perceptron(x1, x2,
                                 params["weights"],
                                 params["bias"])

        print("{:^8} {:^8} {:^15.1f} {:^10}".format(
            x1, x2, net, output))

# -----------------------------
# NOT Gate
# -----------------------------

print("\n" + "=" * 72)
print("NOT GATE")
print("=" * 72)

weight = -1
bias = 0.5

print(f"Weight = {weight}    Bias = {bias}\n")

print("{:^8} {:^15} {:^10}".format(
    "Input", "Weighted Sum", "Output"))
print("-" * 38)

for x in [0, 1]:
    net = x * weight + bias
    output = activation(net)

    print("{:^8} {:^15.1f} {:^10}".format(
        x, net, output))

print("\n" + "=" * 72)
print("Conclusion:")
print("A single-layer perceptron successfully implements")
print("AND, OR, NOT, NAND and NOR gates because they are")
print("linearly separable.")
print("=" * 72)
```
