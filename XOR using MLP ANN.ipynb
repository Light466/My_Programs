```python
# ============================================
# XOR Gate using Multilayer Perceptron (MLP)
# ============================================

# Binary Step Activation Function
def activation(net):
    if net >= 0:
        return 1
    else:
        return 0


# Perceptron Function
def perceptron(x1, x2, weights, bias):
    net = x1 * weights[0] + x2 * weights[1] + bias
    output = activation(net)
    return net, output


print("=" * 70)
print("        XOR GATE USING MULTILAYER PERCEPTRON")
print("=" * 70)

# All possible inputs
inputs = [
    (0, 0),
    (0, 1),
    (1, 0),
    (1, 1)
]

# Hidden Layer Parameters
or_weights = [1, 1]
or_bias = -0.5

nand_weights = [-1, -1]
nand_bias = 1.5

# Output Layer Parameters (AND)
and_weights = [1, 1]
and_bias = -1.5

print("\n{:^8} {:^8} {:^8} {:^8} {:^10}".format(
    "X1", "X2", "OR", "NAND", "XOR"))
print("-" * 48)

for x1, x2 in inputs:

    # Hidden Layer
    _, h1 = perceptron(x1, x2, or_weights, or_bias)
    _, h2 = perceptron(x1, x2, nand_weights, nand_bias)

    # Output Layer
    _, xor = perceptron(h1, h2, and_weights, and_bias)

    print("{:^8} {:^8} {:^8} {:^8} {:^10}".format(
        x1, x2, h1, h2, xor))

print("\n" + "=" * 70)
print("XOR Gate implemented successfully using")
print("a Multilayer Perceptron (2-2-1 Architecture)")
print("=" * 70)
```
