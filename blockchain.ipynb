```python
# ================================================================
# DYNAMIC BLOCKCHAIN-BASED SECURE HOSPITAL EHR
# 100% GOOGLE COLAB COMPATIBLE
#
# Browser GUI using IPyWidgets
# No Tkinter
#
# Features:
#   1. Secure Login
#   2. SHA-256 Password Hashing
#   3. Dynamic 256-bit Session Key
#   4. Dynamic Encryption Key
#   5. Dynamic Authentication Token
#   6. Blockchain Authentication
#   7. Blockchain EHR Records
#   8. Audit Log
#   9. Medicine Repository
#  10. Symptom Repository
#  11. Medicine Sentiment
#  12. Patient Registration
#  13. CSV Generation
#  14. Blockchain Viewer
#
# GOOGLE COLAB VERSION
# ================================================================


# ================================================================
# IMPORT LIBRARIES
# ================================================================

import os
import json
import csv
import hashlib
import secrets
import base64
import datetime
import time
import io

import pandas as pd
import ipywidgets as widgets

from IPython.display import (
    display,
    clear_output,
    HTML
)

from google.colab import files


# ================================================================
# COLAB FILE PATHS
# ================================================================

BASE_DIR = "/content"

BLOCKCHAIN_FILE = os.path.join(
    BASE_DIR,
    "blockchain_db.json"
)

AUDIT_FILE = os.path.join(
    BASE_DIR,
    "audit_log.json"
)

PATIENT_CSV_FILE = os.path.join(
    BASE_DIR,
    "patient_records.csv"
)


# ================================================================
# USER AUTHENTICATION DATABASE
# ================================================================

USERS = {

    "doctor1":
        hashlib.sha256(
            "Doctor@123".encode()
        ).hexdigest(),

    "admin":
        hashlib.sha256(
            "Admin@123".encode()
        ).hexdigest()
}


# ================================================================
# PASSWORD HASHING
# ================================================================

def hash_password(password):

    return hashlib.sha256(
        password.encode("utf-8")
    ).hexdigest()


# ================================================================
# INITIALIZE BLOCKCHAIN
# ================================================================

def initialize_blockchain():

    if not os.path.exists(
        BLOCKCHAIN_FILE
    ):

        genesis_data = {
            "message":
                "Dynamic Secure Hospital EHR Genesis Block",

            "created":
                str(
                    datetime.datetime.now()
                )
        }

        timestamp = str(
            datetime.datetime.now()
        )

        previous_hash = "0"

        hash_input = (
            previous_hash +
            json.dumps(
                genesis_data,
                sort_keys=True
            ) +
            timestamp
        )

        genesis_hash = hashlib.sha256(
            hash_input.encode("utf-8")
        ).hexdigest()

        genesis_block = {

            "index": 0,

            "data":
                genesis_data,

            "timestamp":
                timestamp,

            "previous_hash":
                previous_hash,

            "hash":
                genesis_hash
        }

        with open(
            BLOCKCHAIN_FILE,
            "w",
            encoding="utf-8"
        ) as f:

            json.dump(
                [genesis_block],
                f,
                indent=4
            )


# ================================================================
# LOAD BLOCKCHAIN
# ================================================================

def load_blockchain():

    initialize_blockchain()

    try:

        with open(
            BLOCKCHAIN_FILE,
            "r",
            encoding="utf-8"
        ) as f:

            return json.load(f)

    except Exception:

        return []


# ================================================================
# SAVE BLOCKCHAIN
# ================================================================

def save_blockchain(blockchain):

    with open(
        BLOCKCHAIN_FILE,
        "w",
        encoding="utf-8"
    ) as f:

        json.dump(
            blockchain,
            f,
            indent=4,
            ensure_ascii=False
        )


# ================================================================
# CREATE BLOCK
# ================================================================

def create_block(block_data):

    blockchain = load_blockchain()

    if len(blockchain) == 0:

        previous_hash = "0"

    else:

        previous_hash = blockchain[-1]["hash"]


    timestamp = str(
        datetime.datetime.now()
    )


    data_string = json.dumps(

        block_data,

        sort_keys=True,

        ensure_ascii=False
    )


    hash_input = (

        previous_hash +

        data_string +

        timestamp
    )


    current_hash = hashlib.sha256(

        hash_input.encode("utf-8")

    ).hexdigest()


    new_block = {

        "index":
            len(blockchain),

        "data":
            block_data,

        "timestamp":
            timestamp,

        "previous_hash":
            previous_hash,

        "hash":
            current_hash
    }


    blockchain.append(
        new_block
    )


    save_blockchain(
        blockchain
    )


    return new_block


# ================================================================
# AUDIT LOG
# ================================================================

def load_audit_log():

    if not os.path.exists(
        AUDIT_FILE
    ):

        return []


    try:

        with open(
            AUDIT_FILE,
            "r",
            encoding="utf-8"
        ) as f:

            return json.load(f)

    except Exception:

        return []


def write_audit_log(
        username,
        action
):

    logs = load_audit_log()


    logs.append({

        "user":
            username,

        "action":
            action,

        "time":
            time.time(),

        "datetime":
            str(
                datetime.datetime.now()
            )
    })


    with open(
        AUDIT_FILE,
        "w",
        encoding="utf-8"
    ) as f:

        json.dump(
            logs,
            f,
            indent=4
        )


# ================================================================
# DYNAMIC 256-BIT SESSION KEY
# ================================================================

def generate_session_key():

    # 32 bytes = 256 bits

    return secrets.token_bytes(
        32
    )


# ================================================================
# DYNAMIC ENCRYPTION KEY
# ================================================================

def generate_encryption_key(
        session_key
):

    return base64.urlsafe_b64encode(

        session_key

    ).decode(
        "utf-8"
    )


# ================================================================
# DYNAMIC TOKEN
# ================================================================

def generate_token(
        username,
        session_key
):

    nonce = secrets.token_bytes(
        32
    )


    timestamp = str(
        datetime.datetime.now()
    )


    token_data = (

        username +

        timestamp +

        nonce.hex()
    )


    token = hashlib.sha256(

        session_key +

        token_data.encode(
            "utf-8"
        )

    ).hexdigest()


    return token


# ================================================================
# EHR DEMONSTRATION ENCRYPTION
# ================================================================

def encrypt_ehr(
        ehr_text,
        session_key
):

    ehr_bytes = ehr_text.encode(
        "utf-8"
    )


    encrypted = bytes(

        [

            ehr_bytes[i]

            ^

            session_key[
                i % len(session_key)
            ]

            for i in range(
                len(ehr_bytes)
            )

        ]

    )


    return base64.b64encode(

        encrypted

    ).decode(
        "utf-8"
    )


# ================================================================
# UPLOAD DATA FILES
# ================================================================

upload_title = widgets.HTML(

    value="""
    <h3>Step 1 — Upload Clinical Repository Files</h3>
    <p>
    Upload the following files if available:
    </p>
    <ul>
        <li><b>mydb.Medicines.json</b></li>
        <li><b>mydb.Symptoms.json</b></li>
        <li><b>medicines_sentiment_clean.csv</b></li>
    </ul>
    """
)


upload_button = widgets.Button(

    description="Upload Clinical Files",

    button_style="info",

    icon="upload"
)


upload_output = widgets.Output()


def upload_files():

    with upload_output:

        clear_output()

        print(
            "Select your JSON/CSV clinical repository files."
        )

        uploaded = files.upload()

        if uploaded:

            print()

            print(
                "Uploaded files:"
            )

            for filename in uploaded.keys():

                print(
                    "✓",
                    filename
                )


upload_button.on_click(
    lambda b: upload_files()
)


display(
    upload_title,
    upload_button,
    upload_output
)


# ================================================================
# LOAD MEDICINE DATABASE
# ================================================================

def load_medicine_database():

    possible_files = [

        "/content/mydb.Medicines.json",

        "/content/Medicines.json",

        "/content/medicines.json"
    ]


    for path in possible_files:

        if os.path.exists(path):

            try:

                with open(
                    path,
                    encoding="utf-8"
                ) as f:

                    return json.load(f)

            except Exception:

                pass


    return []


# ================================================================
# LOAD SYMPTOM DATABASE
# ================================================================

def load_symptom_database():

    possible_files = [

        "/content/mydb.Symptoms.json",

        "/content/Symptoms.json",

        "/content/symptoms.json"
    ]


    for path in possible_files:

        if os.path.exists(path):

            try:

                with open(
                    path,
                    encoding="utf-8"
                ) as f:

                    return json.load(f)

            except Exception:

                pass


    return []


# ================================================================
# LOAD SENTIMENT DATABASE
# ================================================================

def load_sentiment_database():

    possible_files = [

        "/content/medicines_sentiment_multilevel.csv"
    ]


    for path in possible_files:

        if os.path.exists(path):

            try:

                df = pd.read_csv(
                    path
                )

                df.columns = (
                    df.columns
                    .str.strip()
                )

                sentiment = {}


                for _, row in df.iterrows():

                    if "Name" not in row:

                        continue


                    name = str(
                        row["Name"]
                    ).strip().lower()


                    sentiment[name] = {

                        "Score":
                            str(
                                row.get(
                                    "SentimentScore",
                                    ""
                                )
                            ),

                        "Label":
                            str(
                                row.get(
                                    "SentimentLabel",
                                    ""
                                )
                            )
                    }


                return sentiment


            except Exception as e:

                print(
                    "Sentiment loading error:",
                    e
                )


    return {}


# ================================================================
# LOAD ALL CLINICAL DATA
# ================================================================

medicines_db = load_medicine_database()

symptoms_db = load_symptom_database()

sentiment_data = load_sentiment_database()


# ================================================================
# PREPARE MEDICINE INFORMATION
# ================================================================

medicine_info = {}


for medicine in medicines_db:

    if "Name" in medicine:

        name = str(
            medicine["Name"]
        ).strip()

        medicine_info[name] = medicine


medicine_names = sorted(
    medicine_info.keys()
)


# ================================================================
# PREPARE SYMPTOM INFORMATION
# ================================================================

symptom_info = {}


for symptom_record in symptoms_db:

    symptoms_field = str(

        symptom_record.get(
            "Symptoms",
            ""
        )
    )


    remedies_text = str(

        symptom_record.get(
            "Remedies",
            ""
        )
    )


    if symptoms_field:

        split_symptoms = [

            s.strip()

            for s in
            symptoms_field.split(",")

        ]


        for symptom in split_symptoms:

            if symptom:

                if symptom not in symptom_info:

                    symptom_info[
                        symptom
                    ] = []


                if remedies_text:

                    symptom_info[
                        symptom
                    ].append(
                        remedies_text
                    )


symptom_names = sorted(
    symptom_info.keys()
)


# ================================================================
# GLOBAL SESSION
# ================================================================

CURRENT_USER = None

CURRENT_SESSION_KEY = None

CURRENT_ENCRYPTION_KEY = None

CURRENT_TOKEN = None


# ================================================================
# LOGIN WIDGETS
# ================================================================

login_title = widgets.HTML(

    value="""
    <div style="
        background:#1f4e78;
        color:white;
        padding:18px;
        border-radius:10px;
        text-align:center;
    ">

        <h2>🔐 SECURE HOSPITAL SYSTEM</h2>

        <p>
        Dynamic Blockchain-Based Secure EHR
        </p>

    </div>
    """
)


username_widget = widgets.Text(

    description="Username:",

    placeholder="doctor1",

    layout=widgets.Layout(
        width="400px"
    )
)


password_widget = widgets.Password(

    description="Password:",

    placeholder="Enter password",

    layout=widgets.Layout(
        width="400px"
    )
)


login_button = widgets.Button(

    description="LOGIN",

    button_style="success",

    icon="sign-in",

    layout=widgets.Layout(
        width="200px"
    )
)


login_output = widgets.Output()


# ================================================================
# LOGIN FUNCTION
# ================================================================

def secure_login(button):

    global CURRENT_USER

    global CURRENT_SESSION_KEY

    global CURRENT_ENCRYPTION_KEY

    global CURRENT_TOKEN


    with login_output:

        clear_output()


        username = (
            username_widget.value
            .strip()
        )


        password = (
            password_widget.value
        )


        # --------------------------------------------------------
        # VERIFY USER
        # --------------------------------------------------------

        if username not in USERS:

            write_audit_log(

                username
                if username
                else "Unknown",

                "Failed Login"
            )


            print(
                "❌ Invalid Username or Password"
            )

            return


        entered_hash = hash_password(
            password
        )


        if entered_hash != USERS[username]:

            write_audit_log(

                username,

                "Failed Login"
            )


            print(
                "❌ Invalid Username or Password"
            )

            return


        # --------------------------------------------------------
        # CREATE DYNAMIC SESSION KEY
        # --------------------------------------------------------

        session_key = (
            generate_session_key()
        )


        # --------------------------------------------------------
        # CREATE DYNAMIC ENCRYPTION KEY
        # --------------------------------------------------------

        encryption_key = (

            generate_encryption_key(
                session_key
            )
        )


        # --------------------------------------------------------
        # CREATE DYNAMIC TOKEN
        # --------------------------------------------------------

        token = generate_token(

            username,

            session_key
        )


        CURRENT_USER = username

        CURRENT_SESSION_KEY = (
            session_key
        )

        CURRENT_ENCRYPTION_KEY = (
            encryption_key
        )

        CURRENT_TOKEN = token


        # --------------------------------------------------------
        # TOKEN FINGERPRINT
        # --------------------------------------------------------

        token_fingerprint = (

            hashlib.sha256(

                token.encode(
                    "utf-8"
                )

            ).hexdigest()
        )


        # --------------------------------------------------------
        # CREATE BLOCKCHAIN LOGIN BLOCK
        # --------------------------------------------------------

        login_block = create_block({

            "type":
                "USER_LOGIN",

            "username":
                username,

            "authentication":
                "SHA-256",

            "session_key_size":
                "256-bit",

            "token_fingerprint":
                token_fingerprint,

            "login_time":
                str(
                    datetime.datetime.now()
                )
        })


        # --------------------------------------------------------
        # AUDIT LOG
        # --------------------------------------------------------

        write_audit_log(

            username,

            "Successful Login"
        )


        # --------------------------------------------------------
        # DISPLAY SECURITY INFORMATION
        # --------------------------------------------------------

        print(
            "=" * 70
        )

        print(
            "✅ LOGIN SUCCESSFUL"
        )

        print(
            "=" * 70
        )

        print(
            "Username:"
        )

        print(
            username
        )

        print()

        print(
            "SHA-256 Password Hash:"
        )

        print(
            USERS[username]
        )

        print()

        print(
            "Dynamic 256-bit Session Key:"
        )

        print(
            session_key.hex()
        )

        print()

        print(
            "Dynamic Encryption Key:"
        )

        print(
            encryption_key
        )

        print()

        print(
            "Dynamic Authentication Token:"
        )

        print(
            token
        )

        print()

        print(
            "Blockchain Block Hash:"
        )

        print(
            login_block["hash"]
        )

        print(
            "=" * 70
        )


        # --------------------------------------------------------
        # SHOW HOSPITAL APPLICATION
        # --------------------------------------------------------

        display_hospital_application()


login_button.on_click(
    secure_login
)


display(

    login_title,

    widgets.VBox([

        username_widget,

        password_widget,

        login_button

    ]),

    login_output
)


# ================================================================
# HOSPITAL APPLICATION
# ================================================================

def display_hospital_application():

    clear_output(
        wait=True
    )


    # ============================================================
    # HOSPITAL HEADER
    # ============================================================

    display(

        HTML(

            f"""
            <div style="
                background:#1f4e78;
                color:white;
                padding:20px;
                border-radius:10px;
                text-align:center;
            ">

            <h1>
            🏥 SECURE HOSPITAL MANAGEMENT SYSTEM
            </h1>

            <p>
            Logged in as:
            <b>{CURRENT_USER}</b>
            </p>

            <p>
            🔐 SHA-256 |
            🔑 256-bit Dynamic Session |
            ⛓ Blockchain Enabled
            </p>

            </div>
            """
        )
    )


    # ============================================================
    # PATIENT VARIABLES
    # ============================================================

    selected_medicines = []

    selected_symptoms = []

    patient_records = []


    # ============================================================
    # PATIENT INPUT WIDGETS
    # ============================================================

    entry_name = widgets.Text(

        description="Patient Name:",

        layout=widgets.Layout(
            width="500px"
        )
    )


    entry_age = widgets.Text(

        description="Age:",

        layout=widgets.Layout(
            width="400px"
        )
    )


    entry_contact = widgets.Text(

        description="Contact:",

        layout=widgets.Layout(
            width="500px"
        )
    )


    entry_department = widgets.Text(

        description="Department:",

        layout=widgets.Layout(
            width="500px"
        )
    )


    combo_gender = widgets.Dropdown(

        description="Gender:",

        options=[
            "",
            "Male",
            "Female",
            "Other"
        ],

        layout=widgets.Layout(
            width="400px"
        )
    )


    entry_date = widgets.Text(

        description="Admission Date:",

        placeholder="DD-MM-YYYY",

        layout=widgets.Layout(
            width="500px"
        )
    )


    entry_doctor = widgets.Text(

        description="Doctor Name:",

        layout=widgets.Layout(
            width="500px"
        )
    )


    # ============================================================
    # MEDICINE WIDGETS
    # ============================================================

    if medicine_names:

        combo_medicine = widgets.Dropdown(

            description="Medicine:",

            options=[
                ""
            ] + medicine_names,

            layout=widgets.Layout(
                width="600px"
            )
        )

    else:

        combo_medicine = widgets.Text(

            description="Medicine:",

            placeholder="Enter medicine",

            layout=widgets.Layout(
                width="600px"
            )
        )


    medicine_add_button = widgets.Button(

        description="Add Medicine",

        button_style="success"
    )


    medicine_remove_button = widgets.Button(

        description="Remove Medicine",

        button_style="warning"
    )


    medicine_list = widgets.SelectMultiple(

        options=[],

        description="Selected:",

        layout=widgets.Layout(
            width="600px",
            height="100px"
        )
    )


    entry_dosage = widgets.Text(

        description="Dosage:",

        layout=widgets.Layout(
            width="500px"
        )
    )


    entry_side_effects = widgets.Textarea(

        description="Side-effects:",

        layout=widgets.Layout(
            width="600px",
            height="80px"
        )
    )


    entry_sentiment_score = widgets.Text(

        description="Sentiment Score:",

        disabled=True,

        layout=widgets.Layout(
            width="450px"
        )
    )


    entry_sentiment_label = widgets.Text(

        description="Sentiment Label:",

        disabled=True,

        layout=widgets.Layout(
            width="500px"
        )
    )


    # ============================================================
    # MEDICINE AUTO-FILL
    # ============================================================

    def auto_fill_medicine(change):

        if change["name"] != "value":

            return


        med = str(
            change["new"]
        ).strip()


        entry_dosage.value = ""

        entry_side_effects.value = ""

        entry_sentiment_score.value = ""

        entry_sentiment_label.value = ""


        if med in medicine_info:

            entry_dosage.value = str(

                medicine_info[
                    med
                ].get(
                    "Dosage",
                    ""
                )
            )


            entry_side_effects.value = str(

                medicine_info[
                    med
                ].get(
                    "Side-effects",
                    ""
                )
            )


        key = med.lower()


        if key in sentiment_data:

            entry_sentiment_score.value = (

                sentiment_data[key][
                    "Score"
                ]
            )


            entry_sentiment_label.value = (

                sentiment_data[key][
                    "Label"
                ]
            )


    combo_medicine.observe(

        auto_fill_medicine,

        names="value"
    )


    # ============================================================
    # ADD MEDICINE
    # ============================================================

    def add_medicine(button):

        med = str(
            combo_medicine.value
        ).strip()


        if med and med not in selected_medicines:

            selected_medicines.append(
                med
            )

            medicine_list.options = (
                tuple(
                    selected_medicines
                )
            )


    medicine_add_button.on_click(
        add_medicine
    )


    # ============================================================
    # REMOVE MEDICINE
    # ============================================================

    def remove_medicine(button):

        selected = list(
            medicine_list.value
        )


        for med in selected:

            if med in selected_medicines:

                selected_medicines.remove(
                    med
                )


        medicine_list.options = (
            tuple(
                selected_medicines
            )
        )


    medicine_remove_button.on_click(
        remove_medicine
    )


    # ============================================================
    # SYMPTOM WIDGETS
    # ============================================================

    if symptom_names:

        combo_symptoms = widgets.Dropdown(

            description="Symptoms:",

            options=[
                ""
            ] + symptom_names,

            layout=widgets.Layout(
                width="600px"
            )
        )

    else:

        combo_symptoms = widgets.Text(

            description="Symptoms:",

            placeholder="Enter symptom",

            layout=widgets.Layout(
                width="600px"
            )
        )


    symptom_add_button = widgets.Button(

        description="Add Symptom",

        button_style="success"
    )


    symptom_remove_button = widgets.Button(

        description="Remove Symptom",

        button_style="warning"
    )


    symptom_list = widgets.SelectMultiple(

        options=[],

        description="Selected:",

        layout=widgets.Layout(
            width="600px",
            height="100px"
        )
    )


    text_remedies = widgets.Textarea(

        description="Remedies:",

        layout=widgets.Layout(
            width="700px",
            height="120px"
        )
    )


    # ============================================================
    # SYMPTOM AUTO-FILL
    # ============================================================

    def auto_fill_symptom(change):

        if change["name"] != "value":

            return


        symptom = str(
            change["new"]
        ).strip()


        text_remedies.value = ""


        if symptom in symptom_info:

            remedies = set()


            for block in symptom_info[
                symptom
            ]:

                parts = [

                    p.strip()

                    for p in block.split(",")

                ]


                for part in parts:

                    if part:

                        remedies.add(
                            part
                        )


            text_remedies.value = (
                "\n".join(
                    sorted(remedies)
                )
            )


    combo_symptoms.observe(

        auto_fill_symptom,

        names="value"
    )


    # ============================================================
    # ADD SYMPTOM
    # ============================================================

    def add_symptom(button):

        symptom = str(
            combo_symptoms.value
        ).strip()


        if symptom and symptom not in selected_symptoms:

            selected_symptoms.append(
                symptom
            )

            symptom_list.options = (
                tuple(
                    selected_symptoms
                )
            )


    symptom_add_button.on_click(
        add_symptom
    )


    # ============================================================
    # REMOVE SYMPTOM
    # ============================================================

    def remove_symptom(button):

        selected = list(
            symptom_list.value
        )


        for symptom in selected:

            if symptom in selected_symptoms:

                selected_symptoms.remove(
                    symptom
                )


        symptom_list.options = (
            tuple(
                selected_symptoms
            )
        )


    symptom_remove_button.on_click(
        remove_symptom
    )


    # ============================================================
    # OUTPUT TABLE
    # ============================================================

    display_columns = [

        "SL No",

        "Patient Name",

        "Age",

        "Contact",

        "Department",

        "Gender",

        "Admission Date",

        "Doctor Name",

        "Symptoms",

        "Medicine",

        "Dosage",

        "Side-effects",

        "Remedies",

        "Medicine Sentiment Score",

        "Medicine Sentiment Label",

        "EHR Hash",

        "Blockchain Hash"
    ]


    output = widgets.Output()


    # ============================================================
    # SAVE PATIENT
    # ============================================================

    save_button = widgets.Button(

        description="SAVE PATIENT RECORD",

        button_style="success",

        icon="save"
    )


    def save_patient(button):

        if not entry_name.value.strip():

            with output:

                print(
                    "❌ Patient Name is required."
                )

            return


        # --------------------------------------------------------
        # CREATE PATIENT RECORD
        # --------------------------------------------------------

        patient_record = {

            "SL No":
                len(patient_records) + 1,

            "Patient Name":
                entry_name.value,

            "Age":
                entry_age.value,

            "Contact":
                entry_contact.value,

            "Department":
                entry_department.value,

            "Gender":
                combo_gender.value,

            "Admission Date":
                entry_date.value,

            "Doctor Name":
                entry_doctor.value,

            "Symptoms":
                ", ".join(
                    selected_symptoms
                ),

            "Medicine":
                ", ".join(
                    selected_medicines
                ),

            "Dosage":
                entry_dosage.value,

            "Side-effects":
                entry_side_effects.value,

            "Remedies":
                text_remedies.value,

            "Medicine Sentiment Score":
                entry_sentiment_score.value,

            "Medicine Sentiment Label":
                entry_sentiment_label.value
        }


        # --------------------------------------------------------
        # CREATE EHR HASH
        # --------------------------------------------------------

        ehr_string = json.dumps(

            patient_record,

            sort_keys=True,

            ensure_ascii=False
        )


        ehr_hash = hashlib.sha256(

            ehr_string.encode(
                "utf-8"
            )

        ).hexdigest()


        # --------------------------------------------------------
        # ENCRYPTION DEMONSTRATION
        # --------------------------------------------------------

        ehr_text = json.dumps(

            patient_record,

            ensure_ascii=False
        )


        encrypted_ehr = encrypt_ehr(

            ehr_text,

            CURRENT_SESSION_KEY
        )


        # --------------------------------------------------------
        # BLOCKCHAIN RECORD
        # --------------------------------------------------------

        block_data = {

            "type":
                "PATIENT_EHR",

            "patient_name":
                entry_name.value,

            "ehr_hash":
                ehr_hash,

            "encrypted_ehr_hash":
                hashlib.sha256(

                    encrypted_ehr.encode(
                        "utf-8"
                    )

                ).hexdigest(),

            "created_by":
                CURRENT_USER,

            "timestamp":
                str(
                    datetime.datetime.now()
                )
        }


        block = create_block(
            block_data
        )


        patient_record[
            "EHR Hash"
        ] = ehr_hash


        patient_record[
            "Blockchain Hash"
        ] = block["hash"]


        patient_records.append(
            patient_record
        )


        # --------------------------------------------------------
        # AUDIT LOG
        # --------------------------------------------------------

        write_audit_log(

            CURRENT_USER,

            "Added Patient: " +
            entry_name.value
        )


        # --------------------------------------------------------
        # OUTPUT
        # --------------------------------------------------------

        with output:

            clear_output()

            print(
                "=" * 80
            )

            print(
                "✅ PATIENT EHR SAVED"
            )

            print(
                "=" * 80
            )

            print(
                "Patient:",
                entry_name.value
            )

            print()

            print(
                "EHR SHA-256 Hash:"
            )

            print(
                ehr_hash
            )

            print()

            print(
                "Blockchain Block:"
            )

            print(
                block["index"]
            )

            print()

            print(
                "Blockchain Hash:"
            )

            print(
                block["hash"]
            )

            print()

            print(
                "Previous Hash:"
            )

            print(
                block["previous_hash"]
            )

            print(
                "=" * 80
            )


        # --------------------------------------------------------
        # CLEAR PATIENT FIELDS
        # --------------------------------------------------------

        entry_name.value = ""

        entry_age.value = ""

        entry_contact.value = ""

        entry_department.value = ""

        combo_gender.value = ""

        entry_date.value = ""

        entry_doctor.value = ""

        combo_medicine.value = ""

        combo_symptoms.value = ""

        entry_dosage.value = ""

        entry_side_effects.value = ""

        entry_sentiment_score.value = ""

        entry_sentiment_label.value = ""

        text_remedies.value = ""

        selected_medicines.clear()

        selected_symptoms.clear()

        medicine_list.options = ()

        symptom_list.options = ()


    save_button.on_click(
        save_patient
    )


    # ============================================================
    # GENERATE CSV
    # ============================================================

    csv_button = widgets.Button(

        description="GENERATE CSV",

        button_style="info",

        icon="download"
    )


    csv_output = widgets.Output()


    def generate_csv(button):

        if not patient_records:

            with csv_output:

                clear_output()

                print(
                    "❌ No patient records available."
                )

            return


        df = pd.DataFrame(
            patient_records
        )


        df.to_csv(

            PATIENT_CSV_FILE,

            index=False,

            encoding="utf-8"
        )


        with csv_output:

            clear_output()

            print(
                "✅ CSV generated:"
            )

            print(
                PATIENT_CSV_FILE
            )


        files.download(
            PATIENT_CSV_FILE
        )


    csv_button.on_click(
        generate_csv
    )


    # ============================================================
    # VIEW BLOCKCHAIN
    # ============================================================

    blockchain_button = widgets.Button(

        description="VIEW BLOCKCHAIN",

        button_style="warning",

        icon="link"
    )


    blockchain_output = widgets.Output()


    def view_blockchain(button):

        blockchain = load_blockchain()


        with blockchain_output:

            clear_output()


            print(
                "=" * 100
            )

            print(
                "DYNAMIC BLOCKCHAIN"
            )

            print(
                "=" * 100
            )


            for block in blockchain:

                print()

                print(
                    "Block Index:",
                    block["index"]
                )

                print(
                    "Timestamp:",
                    block["timestamp"]
                )

                print(
                    "Previous Hash:",
                    block["previous_hash"]
                )

                print(
                    "Current Hash:",
                    block["hash"]
                )

                print(
                    "Data:"
                )

                print(

                    json.dumps(

                        block["data"],

                        indent=4,

                        ensure_ascii=False
                    )

                )

                print(
                    "-" * 100
                )


    blockchain_button.on_click(
        view_blockchain
    )


    # ============================================================
    # VIEW AUDIT LOG
    # ============================================================

    audit_button = widgets.Button(

        description="VIEW AUDIT LOG",

        button_style="warning",

        icon="history"
    )


    audit_output = widgets.Output()


    def view_audit(button):

        logs = load_audit_log()


        with audit_output:

            clear_output()


            print(
                "=" * 80
            )

            print(
                "SECURITY AUDIT LOG"
            )

            print(
                "=" * 80
            )


            if not logs:

                print(
                    "No audit records."
                )

                return


            for log in logs:

                print(
                    "User     :",
                    log.get(
                        "user",
                        ""
                    )
                )

                print(
                    "Action   :",
                    log.get(
                        "action",
                        ""
                    )
                )

                print(
                    "Time     :",
                    log.get(
                        "datetime",
                        ""
                    )
                )

                print(
                    "-" * 80
                )


    audit_button.on_click(
        view_audit
    )


    # ============================================================
    # CURRENT SECURITY INFORMATION
    # ============================================================

    security_box = widgets.HTML(

        value=f"""

        <div style="
            background:#eef6ff;
            border:2px solid #1f4e78;
            padding:15px;
            border-radius:8px;
            margin:10px 0;
        ">

        <h3>🔐 Current Security Session</h3>

        <b>User:</b>
        {CURRENT_USER}

        <br><br>

        <b>Authentication:</b>
        SHA-256

        <br>

        <b>Session Key:</b>
        256-bit dynamic

        <br>

        <b>Blockchain:</b>
        Active

        <br>

        <b>Audit Logging:</b>
        Active

        </div>

        """
    )


    # ============================================================
    # MEDICINE CSV UPLOAD
    # ============================================================

    medicine_csv_button = widgets.Button(

        description="UPLOAD MATCHED MEDICINE CSV",

        button_style="info"
    )


    medicine_csv_output = widgets.Output()


    def upload_medicine_csv(button):

        with medicine_csv_output:

            clear_output()

            print(
                "Select matched medicine CSV."
            )

            uploaded = files.upload()


            for filename, data in uploaded.items():

                try:

                    df = pd.read_csv(
                        io.BytesIO(data)
                    )

                    print()

                    print(
                        "Medicine Pattern Matching Result"
                    )

                    display(df)

                except Exception as e:

                    print(
                        "Error:",
                        e
                    )


    medicine_csv_button.on_click(
        upload_medicine_csv
    )


    # ============================================================
    # SYMPTOM CSV UPLOAD
    # ============================================================

    symptom_csv_button = widgets.Button(

        description="UPLOAD MATCHED SYMPTOM CSV",

        button_style="info"
    )


    symptom_csv_output = widgets.Output()


    def upload_symptom_csv(button):

        with symptom_csv_output:

            clear_output()

            print(
                "Select matched symptom CSV."
            )

            uploaded = files.upload()


            for filename, data in uploaded.items():

                try:

                    df = pd.read_csv(
                        io.BytesIO(data)
                    )

                    print()

                    print(
                        "Symptom Pattern Matching Result"
                    )

                    display(df)

                except Exception as e:

                    print(
                        "Error:",
                        e
                    )


    symptom_csv_button.on_click(
        upload_symptom_csv
    )


    # ============================================================
    # LOGOUT
    # ============================================================

    logout_button = widgets.Button(

        description="LOGOUT",

        button_style="danger",

        icon="sign-out"
    )


    logout_output = widgets.Output()


    def logout(button):

        write_audit_log(

            CURRENT_USER,

            "Logout"
        )


        with logout_output:

            clear_output()

            print(
                "Logged out successfully."
            )

            print(
                "Run the login cell again to start a new session."
            )


    logout_button.on_click(
        logout
    )


    # ============================================================
    # DISPLAY PATIENT FORM
    # ============================================================

    patient_form = widgets.VBox([

        widgets.HTML(
            "<h3>👤 Patient Information</h3>"
        ),

        entry_name,

        entry_age,

        entry_contact,

        entry_department,

        combo_gender,

        entry_date,

        entry_doctor,

        widgets.HTML(
            "<h3>💊 Medicine Information</h3>"
        ),

        combo_medicine,

        widgets.HBox([

            medicine_add_button,

            medicine_remove_button

        ]),

        medicine_list,

        entry_dosage,

        entry_side_effects,

        entry_sentiment_score,

        entry_sentiment_label,

        widgets.HTML(
            "<h3>🩺 Symptoms and Remedies</h3>"
        ),

        combo_symptoms,

        widgets.HBox([

            symptom_add_button,

            symptom_remove_button

        ]),

        symptom_list,

        text_remedies,

        widgets.HBox([

            save_button,

            csv_button

        ]),

        output,

        widgets.HTML(
            "<h3>📊 Clinical CSV Tools</h3>"
        ),

        medicine_csv_button,

        medicine_csv_output,

        symptom_csv_button,

        symptom_csv_output,

        widgets.HTML(
            "<h3>⛓ Security and Blockchain</h3>"
        ),

        widgets.HBox([

            blockchain_button,

            audit_button,

            logout_button

        ]),

        security_box,

        blockchain_output,

        audit_output,

        csv_output,

        logout_output

    ])


    display(
        patient_form
    )


# ================================================================
# INITIALIZE
# ================================================================

initialize_blockchain()

print()
print("=" * 70)
print("DYNAMIC BLOCKCHAIN SECURE HOSPITAL EHR SYSTEM")
print("=" * 70)
print()
print("Login credentials:")
print()
print("Doctor:")
print("Username : doctor1")
print("Password : Doctor@123")
print()
print("Administrator:")
print("Username : admin")
print("Password : Admin@123")
print()
print("=" * 70)
```
