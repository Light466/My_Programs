```python
# ==========================================
# Beginner NLP Program
# Sentiment Analysis
# ==========================================

# Install NLTK
!pip install nltk

# Import libraries
import nltk

from nltk.tokenize import word_tokenize
from nltk.corpus import stopwords
from nltk.stem import PorterStemmer

# Download required resources
nltk.download('punkt')
nltk.download('punkt_tab')
nltk.download('stopwords')

# ------------------------------------------
# 1. Input Text
# ------------------------------------------

text = "I love this movie. It is very good and interesting."

print("Original Text:")
print(text)

# ------------------------------------------
# 2. Convert to Lowercase
# ------------------------------------------

text_lower = text.lower()

print("\nLowercase Text:")
print(text_lower)

# ------------------------------------------
# 3. Tokenization
# ------------------------------------------

words = word_tokenize(text_lower)

print("\nTokens:")
print(words)

# ------------------------------------------
# 4. Remove Stop Words
# ------------------------------------------

stop_words = set(stopwords.words('english'))

filtered_words = []

for word in words:

    if word.isalpha() and word not in stop_words:
        filtered_words.append(word)

print("\nAfter Removing Stop Words:")
print(filtered_words)

# ------------------------------------------
# 5. Stemming
# ------------------------------------------

stemmer = PorterStemmer()

stemmed_words = []

for word in filtered_words:
    stemmed_words.append(stemmer.stem(word))

print("\nAfter Stemming:")
print(stemmed_words)

# ------------------------------------------
# 6. Sentiment Analysis
# ------------------------------------------

positive_words = [
    "good",
    "great",
    "excellent",
    "love",
    "amazing",
    "interesting",
    "wonderful",
    "best"
]

negative_words = [
    "bad",
    "poor",
    "hate",
    "worst",
    "boring",
    "terrible",
    "awful"
]

positive_count = 0
negative_count = 0

for word in filtered_words:

    if word in positive_words:
        positive_count += 1

    if word in negative_words:
        negative_count += 1

# ------------------------------------------
# 7. Display Results
# ------------------------------------------

print("\nPositive Words:", positive_count)
print("Negative Words:", negative_count)

if positive_count > negative_count:

    print("Final Sentiment: Positive")

elif negative_count > positive_count:

    print("Final Sentiment: Negative")

else:

    print("Final Sentiment: Neutral")
```
