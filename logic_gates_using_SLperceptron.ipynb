```python
#Perceptron class implementation
```

```python
import numpy as np
import matplotlib.pyplot as plt

# Simple Perceptron
class Perceptron:
    def __init__(self, input_size, lr=0.1):
        self.weights = np.zeros(input_size + 1)  # +1 for bias
        self.lr = lr

    def activation(self, x):
        return 1 if x >= 0 else 0

    def predict(self, x):
        x = np.insert(x, 0, 1)  # add bias
        z = np.dot(self.weights, x)
        return self.activation(z)
    def train(self, X, y, epochs=10):
        for _ in range(epochs):
            for xi, target in zip(X, y):
                xi_bias = np.insert(xi, 0, 1)
                z = np.dot(self.weights, xi_bias)
                y_pred = self.activation(z)
                error = target - y_pred
                self.weights += self.lr * error * xi_bias
```

```python
#Define Gates and Plotting
```

```python
def plot_gate(gate_name, X, y, model):
    # Generate grid for plotting decision boundary
    x1 = np.linspace(-0.1, 1.1, 100)
    x2 = np.linspace(-0.1, 1.1, 100)
    xx, yy = np.meshgrid(x1, x2)
    Z = np.array([model.predict([x, y]) for x, y in zip(xx.ravel(), yy.ravel())])
    Z = Z.reshape(xx.shape)

    # Plot
    plt.contourf(xx, yy, Z, levels=[-1, 0.5, 2], colors=["lightcoral", "lightgreen"], alpha=0.5)
    plt.scatter(X[:,0], X[:,1], c=y, cmap='bwr', edgecolors='k', s=100)
    plt.title(f"{gate_name} Gate")
    plt.xlabel("Input A")
    plt.ylabel("Input B")
    plt.grid(True)
    plt.show()
```

```python
#Train and Visualize Logic Gates
```

```python
# Logic gate inputs
X = np.array([[0, 0], [0, 1], [1, 0], [1, 1]])

# Dictionary of gates and outputs
gates = {
    "AND": np.array([0, 0, 0, 1]),
    "OR":  np.array([0, 1, 1, 1]),
    "NAND":np.array([1, 1, 1, 0]),
    "NOR": np.array([1, 0, 0, 0])
}

# Train and plot each gate
for gate_name, y in gates.items():
    p = Perceptron(input_size=2)
    p.train(X, y, epochs=10)
    plot_gate(gate_name, X, y, p)
```
