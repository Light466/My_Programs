```python
#Import Libraries
import nltk

from nltk.corpus import movie_reviews
from nltk.corpus import stopwords

from nltk.tokenize import word_tokenize

from nltk.classify import NaiveBayesClassifier
from nltk.classify.util import accuracy
```

```python
# Dataset
nltk.download('movie_reviews')
nltk.download('punkt')
nltk.download('punkt_tab')
nltk.download('stopwords')
```

```python
#Explore the Dataset
print("Categories:")
print(movie_reviews.categories())

print("\nNumber of Reviews:")
print(len(movie_reviews.fileids()))
```

```python
print("Positive Reviews:",
      len(movie_reviews.fileids('pos')))

print("Negative Reviews:",
      len(movie_reviews.fileids('neg')))
```

```python
#Display One Review
review_id = movie_reviews.fileids()[0]

words = movie_reviews.words(review_id)

print(words)
```

```python
#Convert the words into a sentence
text = " ".join(words)

print(text)
```

```python
#Create Positive and Negative Documents
documents = []

for category in movie_reviews.categories():

    for fileid in movie_reviews.fileids(category):

        words = movie_reviews.words(fileid)

        documents.append(
            (list(words), category)
        )

print("Total documents:", len(documents))
```

```python
#Create Word List
all_words = []

for words, category in documents:

    for word in words:

        all_words.append(word.lower())
```

```python
print("Total words:", len(all_words))
```

```python
#English stop-word list
stop_words = set(stopwords.words('english'))
```

```python
#remove stop words
filtered_words = []

for word in all_words:

    if word.isalpha() and word not in stop_words:

        filtered_words.append(word)
```

```python
print(filtered_words[:50])
```

```python
#Find Most Common Words

#NLTK provides FreqDist() for calculating word frequency.

from nltk import FreqDist

word_frequency = FreqDist(filtered_words)

print(word_frequency.most_common(20))
```

```python
#Select the Most Important Words

#We will use the 3,000 most frequent words.

most_common_words = [
    word
    for word, frequency in word_frequency.most_common(3000)
]

print("Number of selected words:",
      len(most_common_words))
```

```python
#Create Features

#This is an important part.

#For every review, we check whether an important word is present.

def document_features(document):

    document_words = set(document)

    features = {}

    for word in most_common_words:

        features["contains_" + word] = (
            word in document_words
        )

    return features
```

```python
#create dataset
feature_sets = [
    (document_features(words), category)
    for words, category in documents
]

print("Feature sets:", len(feature_sets))
```

```python
#Split Training and Testing Data

#We will use:80% training, 20% testing
train_set = feature_sets[:1600]

test_set = feature_sets[1600:]

print("Training samples:", len(train_set))
print("Testing samples:", len(test_set))
```

```python
#Train Naive Bayes Classifier
classifier = NaiveBayesClassifier.train(train_set)
```

```python
#Calculate Accuracy
accuracy_score = accuracy(
    classifier,
    test_set
)

print("Accuracy:", accuracy_score)
print("Accuracy Percentage:",
      accuracy_score * 100)
```
