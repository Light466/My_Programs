```python
import pandas as pd
from sklearn import svm
from sklearn.preprocessing import LabelEncoder
from sklearn.metrics import accuracy_score, classification_report
```

```python
# Separate features and target
X_train = df1.drop("prognosis", axis=1)
y_train = df1["prognosis"]

X_test = df1.drop("prognosis", axis=1)
y_test = df1["prognosis"]
```

```python
# Encode target labels
le = LabelEncoder()
y_train_encoded = le.fit_transform(y_train)
y_test_encoded = le.transform(y_test)
```

```python
# Initialize and train SVM classifier
clf = svm.SVC(kernel='linear')  # You can use 'rbf' or 'poly' as well
clf.fit(X_train, y_train_encoded)

# Predict on test data
y_pred = clf.predict(X_test)

# Evaluate the model
print("SVM Classifier Results")
print("Accuracy:", accuracy_score(y_test_encoded, y_pred))
print("Classification Report:")
print(classification_report(y_test_encoded, y_pred, target_names=le.classes_))
```

```python
import pandas as pd
from sklearn.tree import DecisionTreeClassifier
from sklearn.preprocessing import LabelEncoder
from sklearn.metrics import accuracy_score, classification_report
```

```python
# Split features and labels
X_train = df1.drop("prognosis", axis=1)
y_train = df1["prognosis"]

X_test = df1.drop("prognosis", axis=1)
y_test = df1["prognosis"]
```

```python
# Encode disease labels
le = LabelEncoder()
y_train_encoded = le.fit_transform(y_train)
y_test_encoded = le.transform(y_test)

# Initialize and train Decision Tree
clf = DecisionTreeClassifier(criterion='entropy', random_state=0)
clf.fit(X_train, y_train_encoded)

# Predict on test data
y_pred = clf.predict(X_test)

# Evaluate performance
print("🌳 Decision Tree Classifier Results")
print("Accuracy:", accuracy_score(y_test_encoded, y_pred))
print("Classification Report:")
print(classification_report(y_test_encoded, y_pred, target_names=le.classes_))
```

```python
import pandas as pd
from sklearn.naive_bayes import GaussianNB
from sklearn.preprocessing import LabelEncoder
from sklearn.metrics import accuracy_score, classification_report
```

```python
# Split features and labels
X_train = df1.drop("prognosis", axis=1)
y_train = df1["prognosis"]

X_test = df1.drop("prognosis", axis=1)
y_test = df1["prognosis"]

# Encode the target labels (disease names to integers)
le = LabelEncoder()
y_train_encoded = le.fit_transform(y_train)
y_test_encoded = le.transform(y_test)
```

```python
# Initialize and train Naive Bayes classifier
nb_classifier = GaussianNB()
nb_classifier.fit(X_train, y_train_encoded)

# Predict on test set
y_pred = nb_classifier.predict(X_test)

# Evaluate the model
print("Naive Bayes Classifier Results")
print("Accuracy:", accuracy_score(y_test_encoded, y_pred))
print("Classification Report:")
print(classification_report(y_test_encoded, y_pred, target_names=le.classes_))
```
